3. Data Quality Checks

3.1. Duplicate Records

```python
print("Promotions :",Promotions['PromotionID'].duplicated().sum())
print("Customers :",Customers['CustomerID'].duplicated().sum())
print("Products :",Products['ProductID'].duplicated().sum())
print("Payments :",Payments['PaymentID'].duplicated().sum())
print("Stores :",Stores['StoreID'].duplicated().sum())
print("Sales :",Sales['OrderID'].duplicated().sum())
```

3.2. Missing Values

```python
datasets = {
    "Customers": Customers,
    "Sales": Sales,
    "Products": Products,
    "Payments": Payments,
    "Promotions": Promotions,
    "Stores": Stores
}

missing_summary = pd.DataFrame({
    "Rows": {name: df.shape[0] for name, df in datasets.items()},
    "Columns": {name: df.shape[1] for name, df in datasets.items()},
    "Missing Cells": {name: int(df.isna().sum().sum())
                      for name, df in datasets.items()},
    "Missing Percentage": {
        name: round(df.isna().sum().sum() /
                    (df.shape[0] * df.shape[1]) * 100, 2)
        for name, df in datasets.items()
    }
})

missing_summary
```

```python
for name, df in datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]

    print(f"\n{name}")
    if missing.empty:
        print("No missing values")
    else:
        print(missing)
```

4. Data Cleaning and Transformation342

```python
Sales["PromotionID"]=Sales["PromotionID"].fillna("No Promotion")

print("Missing Values after replacement :", Sales["PromotionID"].isna().sum())
```

```python
Customers["SignupDate"] = pd.to_datetime(
    Customers["SignupDate"],
    errors="coerce"
)

print("Invalid Signup Dates:", Customers["SignupDate"].isna().sum())
```

```python
Sales['OrderDateTime'] = pd.to_datetime(
    Sales['OrderDateTime'],
    errors='coerce'
)
print("Invalid Order Dates:",Sales['OrderDateTime'].isna().sum())
```

6. Saving Cleaned CSV Data

```python

# Create a folder for cleaned files
OUTPUT_DIR = Path("cleaned_data")
OUTPUT_DIR.mkdir(exist_ok=True)

# Save all six cleaned datasets
Customers.to_csv(OUTPUT_DIR / "Customers_Information.csv", index=False)
Sales.to_csv(OUTPUT_DIR / "Sales_Data.csv", index=False)
Products.to_csv(OUTPUT_DIR / "Products_Information.csv", index=False)
Payments.to_csv(OUTPUT_DIR / "Payments_Information.csv", index=False)
Promotions.to_csv(OUTPUT_DIR / "Promotions_Information.csv", index=False)
Stores.to_csv(OUTPUT_DIR / "Stores_Information.csv", index=False)

print("All six cleaned CSV files saved successfully!")
```
